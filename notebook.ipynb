```python
Question 1
#1. Regression is used to predict a numerical value. Regression can be used to predict someone's house price or blood pressure. Classification is used when we want to predict a category or class, and it could be used to predict whether an email is spam or not spam. Regression predicts numbers and classification predicts categories.
#2. A confusion table compares the actual class of an observation to the class that the model predicted. It shows true positives, true negatives, false positives, and false negatives, which helps to understand not only how accurate the model is, but also what types of mistakes the model is making.
#3. SSE (Sum of Squared Errors), measures how far the model's predicted values are from the actual values. It adds up the squared differences between the actual and predicted values. A lower SSE means that the model's predictions are closer to the actual values, so the model fits the data better.
#4. Overfitting happens when a model is too complex and learns the training data too closely, including random noise. It can perform well on the training data but poorly on new data. Underfitting happens when a model is too simple and does not capture the important patterns in the data. An underfit model usually performs poorly on both the training and testing data.
#5. Splitting the data allows to train the model on one set of data and then test it on data that the model has not seen before. This helps to see how well the model will perform on new data instead of just seeing how well it memorizes the training data. By comparing accuracy or SSE for different values of k on the test set, we can choose the value of k that gives better performance and avoid overfitting.
#6. Example: The model could predict that a patient has a disease or does not have a disease. A class label does not tell us how confident the model is in its prediction. A probability distribution gives more information because it shows the probability of each possible class. For example, the model could say there is an 80% chance of disease and a 20% chance of no disease. This gives us more information about the model's confidence, but it can be more complicated to interpret and we may still need to choose a class based on the probabilities.
```

```python
#Question 2.1
import pandas as pd
data = pd.read_csv('land_mines.csv')
print(data.head())
print(data.shape)
print(data.describe())
print(data['mine_type'].value_counts().sort_index())

#the The dataset has 338 observations and 4 variables
# mine_type is the target variable
# voltage, height, and soil are features used to predict the type of mine
# There are 5 different mine type
# Mine type 1: 71 observations
# Mine type 2: 70 observations
# Mine type 3: 66 observations
# Mine type 4: 66 observations
# Mine type 5: 65 observations
#The classes are around evenly distributed, there is not a major imbalance in mine_type.
#Features: the mean voltage is about 0.431, the mean height is about 0.509, and the mean soil value is about 0.504.
```

```python
#Question 2.2

from sklearn.model_selection import train_test_split
X = data[["voltage", "height", "soil"]]
y = data["mine_type"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42, stratify=y)

# The data is split into 50% test data and 50% training data.
# Stratification is used so that each type of mine is represented in both datasets
```

```python
#Question 2.3

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in range(1, 11):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    predictions = knn.predict(X_test)
    print(k, accuracy_score(y_test, predictions))

  #I used a k-nearest neighbors (k-NN) classifier to predict the type of mine based on its voltage, height, and soil measurements.
  #k-NN works by looking at the k observations that are closest to a new observation and using their mine types to make a prediction.
  #I tested different values of k and calculated the accuracy for each one.
  #I then chose the value of k with the highest accuracy because it performed the best at predicting the mine type.
  #The average validation accuracy for k=2 was 43.7%, so I used k=2 for the final model.
```

```python
#Question 2.4

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score

knn = KNeighborsClassifier(n_neighbors=2)

knn.fit(X_train, y_train)

predictions = knn.predict(X_test)

print(confusion_matrix(y_test, predictions))

print(accuracy_score(y_test, predictions))

# The accuracy is 43.8%, so the model correctly predicted 43.8% of test observations
# In the matrix, the diagonal shows the correct predictions.
# Each row is a different mine type, and through the matrix, you can see that mine type 2 had the greatest prediction accuracy (32)
# The model had a difficult time predicting the accuracy of mine type 5 (0)
# Overall, the model was good at predicting mine type 1 and mine type 2, and worse at predicting the rest.
```
