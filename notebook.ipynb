```python
#Question 2.1
import pandas as pd
data = pd.read_csv('land_mines.csv')
print(data.head())
print(data.shape)
print(data.describe())
print(data['mine_type'].value_counts().sort_index())

#the The dataset has 338 observations and 4 variables
# mine_type is the target variable
# voltage, height, and soil are features used to predict the type of mine
# There are 5 different mine type
# Mine type 1: 71 observations
# Mine type 2: 70 observations
# Mine type 3: 66 observations
# Mine type 4: 66 observations
# Mine type 5: 65 observations
#The classes are around evenly distributed, there is not a major imbalance in mine_type.
#Features: the mean voltage is about 0.431, the mean height is about 0.509, and the mean soil value is about 0.504.
```

```python
#Question 2.2

from sklearn.model_selection import train_test_split
X = data[["voltage", "height", "soil"]]
y = data["mine_type"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42, stratify=y)

# The data is split into 50% test data and 50% training data.
# Stratification is used so that each type of mine is represented in both datasets
```

```python
#Question 2.3

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in range(1, 11):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    predictions = knn.predict(X_test)
    print(k, accuracy_score(y_test, predictions))

  #I used a k-nearest neighbors (k-NN) classifier to predict the type of mine based on its voltage, height, and soil measurements.
  #k-NN works by looking at the k observations that are closest to a new observation and using their mine types to make a prediction.
  #I tested different values of k and calculated the accuracy for each one.
  #I then chose the value of k with the highest accuracy because it performed the best at predicting the mine type.
  #The average validation accuracy for k=2 was 43.7%, so I used k=2 for the final model.
```

```python
#Question 2.4

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score

knn = KNeighborsClassifier(n_neighbors=2)

knn.fit(X_train, y_train)

predictions = knn.predict(X_test)

print(confusion_matrix(y_test, predictions))

print(accuracy_score(y_test, predictions))

# The accuracy is 43.8%, so the model correctly predicted 43.8% of test observations
# In the matrix, the diagonal shows the correct predictions.
# Each row is a different mine type, and through the matrix, you can see that mine type 2 had the greatest prediction accuracy (32)
# The model had a difficult time predicting the accuracy of mine type 5 (0)
# Overall, the model was good at predicting mine type 1 and mine type 2, and worse at predicting the rest.
```

```python
#Question 2.1
import pandas as pd
data = pd.read_csv('land_mines.csv')
print(data.head())
print(data.shape)
print(data.describe())
print(data.isna().sum())                       # new line
print(data['mine_type'].value_counts().sort_index())
print(data.groupby('mine_type').mean())        # new line

# The dataset has 338 observations and 4 variables, with no missing values.
# mine_type is the target variable
# voltage, height, and soil are features used to predict the type of mine
# There are 5 different mine types
# Mine type 1: 71 observations
# Mine type 2: 70 observations
# Mine type 3: 66 observations
# Mine type 4: 66 observations
# Mine type 5: 65 observations
# The classes are roughly evenly distributed, so there is no major imbalance in mine_type.
# Features: the mean voltage is about 0.431, the mean height is about 0.509, and the mean soil value is about 0.504.
# All three features are already on a similar 0 to 1 scale, so I did not standardize them for k-NN.
# Looking at the means by mine type, voltage differs the most across classes. Mine type 2 has a much higher average
# voltage (about 0.72) than the other types (about 0.30 to 0.40). Height and soil have very similar averages
# across all mine types (around 0.5), so they seem to separate the classes less.
```

```python
#Question 2.2

from sklearn.model_selection import train_test_split
X = data[["voltage", "height", "soil"]]
y = data["mine_type"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42, stratify=y)

# The data is split into 50% training data and 50% test data.
# Stratification is used so that each type of mine is represented in both datasets.
```

```python
#Question 2.3

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in range(1, 11):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    predictions = knn.predict(X_test)
    print(k, accuracy_score(y_test, predictions))

# I used a k-nearest neighbors classifier to predict the type of mine based on its voltage, height, and soil measurements.
# kNN works by looking at the k observations that are closest to a new observation and using their mine types to make a prediction.
# I tested values of k from 1 to 10 and calculated the test set accuracy for each one.
# I then chose the value of k with the highest accuracy because it performed the best at predicting the mine type.
# k = 2 had the highest accuracy (43.8%), so I used k = 2 for the final model.
# Since k was chosen using the same test/validation set, the final accuracy may be slightly optimistic.
# The accuracies for different values of k were also fairly close to each other, so the choice of k is not very precise.
```

```python
#Question 2.4

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score

knn = KNeighborsClassifier(n_neighbors=2)

knn.fit(X_train, y_train)

predictions = knn.predict(X_test)

print(confusion_matrix(y_test, predictions))

print(accuracy_score(y_test, predictions))

# The accuracy is 43.8%, so the model correctly predicted 43.8% of test observations.
# In the confusion matrix, each row is the actual mine type and each column is the predicted mine type,
# so the diagonal shows the correct predictions.
# Performance was best for mine type 2, where 32 of 35 were correct (about 91%).
# Mine type 1 was also fairly good, with 25 of 36 correct (about 69%).
# Performance was much worse for the other types: mine type 3 was correct 9 of 33 times (about 27%),
# mine type 4 was correct 8 of 33 times (about 24%), and mine type 5 was never predicted correctly (0 of 32).
# Many mine types 3, 4, and 5 were wrongly predicted as mine type 1.
# Overall, the model was good at predicting mine types 1 and 2, and much worse at the rest.
```
